```python
import sa
import random
import matplotlib.pyplot as plt
import json
```

```python
with open("../data/packages.json") as f:
    packages = json.load(f)
with open("../data/truck.json") as f:
    truck = json.load(f)
```

```python
for x in (1, 2, 3):
    r = sa.Run(packages, truck, seed=x)
    sa.saveRun(r, packages, truck, f"../../output/simulated_annealing/run{x}")
    print(f"run {x}: final {r['finalVal']}, {r['duration']:.2f} s, stuck {r['stuck']}")
```
